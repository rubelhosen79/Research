```python
from transformers import pipeline

classifier = pipeline(
    "text-classification",
    model="mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis",
    tokenizer="mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis"
)
```

```python
print("Classifying texts...")
results = []

for index, text in enumerate(data):
    prediction = classifier(text[:512])[0]

    # Map labels
    label_map = {'positive': 'positive', 'negative': 'negative', 'neutral': 'neutral'}
    sentiment = label_map.get(prediction['label'].lower(), prediction['label'])

    result = {
        'Entry_Number': index + 1,
        'Text': text,
        'Sentiment': sentiment,

    }
    results.append(result)

    # Display result
    print(f"Entry #{index + 1}")
    print(f"Text: {text}")
    print(f"Classification: {sentiment}")
    print("="*100)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, f1_score
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print("Classifying texts...")
results = []
predictions = []

for index, text in enumerate(data):
    prediction = classifier(text[:512])[0]

    # Map labels
    label_map = {'positive': 'positive', 'negative': 'negative', 'neutral': 'neutral'}
    sentiment = label_map.get(prediction['label'].lower(), prediction['label'])

    result = {
        'Entry_Number': index + 1,
        'Text': text,
        'Sentiment': sentiment,
    }
    results.append(result)
    predictions.append(sentiment)

    # Display result
    print(f"Entry #{index + 1}")
    print(f"Text: {text}")
    print(f"Classification: {sentiment}")
    print("="*100)

# Assuming you have true_labels list with ground truth
# true_labels = [...]  # Your actual labels

# Calculate metrics
print("\n" + "="*100)
print("EVALUATION METRICS")
print("="*100)

# Confusion Matrix
cm = confusion_matrix(true_labels, predictions)
print("\nConfusion Matrix:")
print(cm)

# Visualize Confusion Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['negative', 'neutral', 'positive'],
            yticklabels=['negative', 'neutral', 'positive'])
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.title('Confusion Matrix')
plt.show()

# F1-Score and Classification Report
print("\nClassification Report:")
print(classification_report(true_labels, predictions))

# F1-scores for each class
#f1_macro = f1_score(true_labels, predictions, average='macro')
f1_weighted = f1_score(true_labels, predictions, average='weighted')

#print(f"\nF1-Score (Macro): {f1_macro:.4f}")
print(f"F1-Score (Weighted): {f1_weighted:.4f}")
```
